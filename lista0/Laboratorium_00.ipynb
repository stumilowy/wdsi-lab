## 3. Indeksowanie list
W Pythonie każdy element w liście ma przypisaną swoją pozycję, czyli indeks. Indeksy liczymy od zera, czyli pierwszy element ma indeks 0, drugi 1 ... Możemy również odnosić się do elementów od końca, używając indeksów ujemnych, np. -1 oznacza ostatni element w liście.

Często pobranie jednego elementu nie jest wystarczające i potrzebujemy pobrać z listy zakres. Nazywamy to wycinaniem (z ang. slicing). Używamy zapisu `lista[start:koniec]`, gdzie start to indeks początkowy, a koniec to indeks, na którym kończymy. Zakres ten jest lewostronnie domknięty, czyli `lista[start]` będzie elementem wyniku, ale `lista[koniec]` już nie. Ostatnim elementem wyniku będzie `lista[koniec-1]`.

````python
print("Cała lista owoców:", favorite_fruits)
print("Pierwszy owoc:", favorite_fruits[0])
print("Pierwsze dwa owoce:", favorite_fruits[0:2])
print("Ostatni owoc:", favorite_fruits[-1])
````

## 4. Operacje matematyczne

Python może wykonywać operacje matematyczne. Możesz dodawać (`+`), odejmować (`-`), mnożyć (`*`) i dzielić (`/`) liczby. Dodatkowo, operator `**` pozwala na potęgowanie, np. `2 ** 8 = 256`. Operacje matematyczne są podstawą wielu programów: od prostych obliczeń po gry komputerowe i sztuczną inteligencję.

````python
a = 2
b = 8

print(a, "+", b, "=", a + b)   # dodawanie
print(a, "-", b, "=", a - b)   # odejmowanie
print(a, "*", b, "=", a * b)   # mnożenie
print(a, "/", b, "=", a / b)   # dzielenie
print(a, "**", b, "=", a ** b)  # potęgowanie
print(b, "%", a, "=", b % a)   # reszta z dzielenia
````

Możemy też porównywać ze sobą liczby oraz inne typy zmiennych. Wynikiem takiego porównania jest wartość logiczna mówiąca o prawdziwości twierdzenia.


````python
print(a, "+", 6, ">", b, ":", a + 6 > b)   # a + 6 większe od b
print(a, "+", 6, "<", b, ":", a + 6 < b)   # a + 6 mniejsze od b
print(a, "+", 6, "==", b, ":", a + 6 == b)  # a + 6 równe b
print("kot == pies", ":", "kot" == "pies")  # porównanie ciągów znaków
````

## 5. Instrukcje warunkowe – `if, elif, else`
Programy komputerowe często muszą podejmować decyzje na podstawie danych (np. poleceń użytkownika, ich wewnętrznego stanu). Do sterowania logiką programu służą instrukcje warunkowe. Konstrukcja opiera się na słowach `if` (jeśli), `elif` (skrót od else if, czyli "w przeciwnym razie jeśli") oraz `else` (w przeciwnym razie, co jest prawdziwe dla wszystkich przypadków nieobsłużonych przez `if` i `elif`). Zwróć uwagę na wcięcia w kodzie. Są niezbędne, gdyż informują Python, które polecenia należą do danego warunku.

````python
temperature = 37

if temperature > 36.6:
    print("Masz gorączkę. Powinieneś iść do lekarza.")
elif temperature < 36.6 and temperature > 35.5:
    # Ten blok wykona się, jeśli pierwszy warunek (temperature > 36.6) jest nieprawdziwy (False),
    # ale aktualny (temperature < 36.6 and temperature > 35.5) jest prawdziwy (True).
    print("Masz niską temperaturę. Napij się czegoś ciepłego.")
elif temperature < 35.5:
    # Ten blok wykona się, jeśli oba powyższe warunki są nieprawdziwe (False),
    # ale aktualny (temperature < 35.5) jest prawdziwy (True).
    print("Masz bardzo niską temperaturę. Natychmiast idź do lekarza.")
else:
    # Ten blok wykona się tylko wtedy, gdy żaden z powyższych warunków nie został spełniony.
    print("Twoja temperatura ciała jest prawidłowa.")
````

## 6. Pętla `for` – powtarzanie czynności

W programowaniu często trzeba wykonać tę samą czynność dla wielu elementów. Pętla `for` pozwala to zrobić w prosty sposób. Na przykład: `for fruit in favorite_fruits:` przechodzi przez każdy owoc w liście `favorite_fruits` i go wypisuje. Takie rozwiązanie jest bardzo przydatne, gdy przetwarzasz dane z plików, listy użytkowników albo chcesz coś narysować wiele razy w programie graficznym. Zwróć uwagę na wcięcie w kodzie – w Pythonie wcięcia służą do wyznaczania struktury i bloków kodu. Są niezbędne, by program działał poprawnie, tzn. wykonywał polecenia tylko w ściśle określonych miejscach w kodzie.


````python
# wykorzystujemy wcześniej zdefiniowaną zmienną favorite_fruits
for fruit in favorite_fruits:
    print("Znany owoc:", fruit)

# dodajmy do listy kolejne owoce wykorzystując pętlę for i metodę append
for fruit in ["truskawka", "kiwi", "jagoda"]:
    favorite_fruits.append(fruit)

print("Po dodaniu nowych owoców:", favorite_fruits)
````

Gdybyśmy źle zastosowali wcięcia w kodzie, moglibyśmy uzyskać nieoczekiwany rezultat. Załóżmy, że chcemy wyświetlić liczby od 1 do 5, a następnie od 11 do 15.

Do wygenerowania zakresów liczb wykorzystamy `range` – sekwencję generującą liczby z wybranego zakresu. Aby wygenerować liczby z zakresu od 1 do 5, należy podać 1 jako początek zakresu, a 5+1 jako koniec zakresu. Takie działanie ma swoje praktyczne uzasadnienia, które staną się jasne w toku nauki programowania (nie tylko w Pythonie).


````python
# Wyświetlanie liczb od 1 do 5
for i in range(1, 6):
    print(i)

for i in range(11, 16):
    print(i)
````

Co się stanie, gdy niepoprawnie zastosujemy wcięcia i drugą pętlę stworzymy wewnątrz pierwszej? Cała druga pętla wykona się przy każdym przejściu pierwszej pętli. Dojdzie do tzw. *zagnieżdżenia pętli*, co czasem jest pożądane, jednak nie w tym przypadku.


````python
# Wyświetlanie liczb od 1 do 5
for i in range(1, 6):
    print(i)
# Wyświetlanie liczb od 11 do 15
    for i in range(11, 16):
        print(i)
````

## 9. Zapis i odczyt plików

Wiele programów potrzebuje zapisywać dane lub wczytywać je z plików. Python to umożliwia w prosty sposób. Używając `open("plik.txt", "w")`, możesz utworzyć plik i zapisać w nim dane. Później, dzięki `open("plik.txt", "r")`, można je odczytać. To przydatne w aplikacjach zapisujących przebieg działania programu, ustawienia użytkownika czy analizujących dane przechowywane w plikach.


````python
# otwórz plik do zapisu
with open("sample_file.txt", "w") as f:
    # \n to znak nowej linii - informuje, że tekst będzie zapisany w nowej linii
    # bez niego tekst będzie zapisany w jednej linii
    f.write("-5.5\n")
    f.write("10.0\n")
    f.write("15.9\n")

# otwórz plik do odczytu
with open("sample_file.txt", "r") as f:
    # wczytaj wszystkie linie z pliku
    content = f.readlines()
# wyświetl zawartość pliku
print(content)
````

Ponieważ wartości zostały wczytane z pliku, są one traktowane jako tekst (ciągi znaków). W celu pozbycia się znaków nowej linii możemy użyć metody `.strip`, dostępnej dla każdego ciągu znaków.

Poniżej zastosowano `list comprehension`. Jest to specjalna konstrukcja w języku Python, która umożliwia stworzenie listy w jednej linijce. Alternatywnie można użyć pętli `for`. Efekt będzie taki sam.


````python
stripped_content = [line.strip() for line in content]  # usuń znaki nowej linii
print("Z użyciem list comprehension:", stripped_content)

stripped_content_for = []
for line in content:
    stripped_content_for.append(line.strip())  # usuń znaki nowej linii
print("Z użyciem pętli for:", stripped_content_for)
````

Przeczytane wartości są liczbami, jednak apostrofy wskazują, że widzimy je jako ciągi znaków. Będzie to przeszkadzało w wykonywaniu operacji matematycznych na nich. Np. dodawanie ciągów znaków powoduje ich łączenie.


````python
print("Dodawanie ciągów znaków:", stripped_content[0] + stripped_content[1] + stripped_content[2])
````

Możemy jednak wykonać *rzutowanie* wartości, aby faktycznie stały się liczbami. Naprawi to problematyczne dodawanie.


````python
floats = [float(line) for line in stripped_content]  # rzutowanie na wartości zmiennoprzecinkowe (float)
print("Rzutowanie na float:", floats)
print("Suma wartości:", floats[0] + floats[1] + floats[2])  # suma wartości
print("Suma wartości z użyciem funkcji sum:", sum(floats))  # suma wartości z użyciem funkcji sum
````

## 11. `pandas.DataFrame`

Wygodnym sposobem reprezentacji danych, który zostanie z Tobą najpewniej do końca studiów, jest typ DataFrame (ramka danych) implementowany w bibliotece pandas. Służy do przedstawienia danych w postaci tabeli i umożliwia łatwe filtrowanie, agregację, czy wykonywanie operacji dla wszystkich elementów. Choć typy kolumn nie zawsze są jednoznaczne (generowane jest ostrzeżenie) i zaawansowane metody mogą być trudne w implementacji, DataFrame są często używane w analizie danych.

DataFrame możemy stworzyć na wiele sposobów, jednak jednym z najprostszych jest wykorzystanie słownika, lub listy słowników. Przy korzystaniu z listy słowników, jak w przykładzie poniżej, każdy element listy jest pojedynczym wierszem naszej tabeli.

````python
# Pociągi Kolei Dolnośląskich
# Lista reprezentacji ramowych (słowników) pociągów KD

trains_kd = [
    {
        "manufacturer": "Newag",
        "model": "31WE",
        "year": 2013,
        "is_electric": True,
        "max_speed_kmh": 160
    },
    {
        "manufacturer": "Newag",
        "model": "36WEa",
        "year": 2014,
        "is_electric": True,
        "max_speed_kmh": 160
    },
    {
        "manufacturer": "Newag",
        "model": "45WE",
        "year": 2017,
        "is_electric": True,
        "max_speed_kmh": 160
    },
    {
        "manufacturer": "Pafawag / ZNTK Mińsk Mazowiecki",
        "model": "EN57AL",
        "year": 1992,
        "is_electric": True,
        "max_speed_kmh": 110
    },
    {
        "manufacturer": "Pafawag / ZNTK Mińsk Mazowiecki",
        "model": "EN57AKD",
        "year": 1991,
        "is_electric": True,
        "max_speed_kmh": 110
    },
    {
        "manufacturer": "Pesa SA",
        "model": "48WEc",
        "year": 2022,
        "is_electric": True,
        "max_speed_kmh": 160
    }
]
````

````python
import pandas as pd

# konwersja listy słowników do DataFrame Pandas
df = pd.DataFrame(trains_kd)
df
````

````python
# wyświetl tylko pociągi wyprodukowane po 2010 roku
# df["year"] > 2010 zwraca True lub False dla każdego wiersza, zależnie od wartości w kolumnie "year"
# df[...] zwraca tylko te wiersze, dla których warunek jest spełniony (True)
print(df[df["year"] > 2010])
````

````python
# wyświetl średnią maksymalną prędkość pociągów
print("Średnia maksymalna prędkość pociągów KD:", f"{df['max_speed_kmh'].mean():.2f}", "km/h")
````

````python
# oblicz czas przejazdu 100 km przy maksymalnej prędkości
def time_per_100_km(max_speed_kmh):
    if max_speed_kmh <= 0:
        return None
    return round(100 / max_speed_kmh * 60, 2)  # minutes

# dodaj nową kolumnę z czasem przejazdu 100 km
df["time_per_100km"] = df["max_speed_kmh"].apply(time_per_100_km)
df
````
