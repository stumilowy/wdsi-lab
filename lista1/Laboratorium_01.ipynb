## Zadanie 2

Celem nadal jest dobranie współczynników tak, aby krzywa była jak najlepiej dopasowana do punktów pomiarowych. W tym zadaniu jednak, zamiast dobierać współczynniki ręcznie, zautomatyzujemy proces przez wykorzystanie metody przeglądu zupełnego (ang. *grid search*).

Przed przeszukaniem musimy jednak najpierw określić sposób oceny rozwiązania — metrykę. Pozwoli ona komputerowi porównać dwa rozwiązania i wybrać lepsze z nich na podstawie zaprogramowanej logiki. W tym zadaniu wykorzystamy jedną z najczęściej używanych w sztucznej inteligencji metryk — błąd średniokwadratowy.

Błąd średniokwadratowy (*Mean Squared Error*, MSE) mierzy jakość dopasowania modelu do danych. W zadaniu 1 to właśnie MSE było raportowane jako błąd dopasowania. Intuicyjnie działa podobnie jak ocena „na oko” — bierze pod uwagę różnice między punktami pomiarowymi a wartościami przewidywanymi przez model. MSE to średnia wartość kwadratów różnic między wartościami rzeczywistymi (lub zmierzonymi) a przewidywanymi:

$$MSE = {1 \over n} \sum_{i=1}^n (y_i - \hat{y}_i)^2$$

Gdzie:

$n$ – liczba próbek (obserwacji),

$y_i$ – rzeczywista (zmierzona / obserwowana) wartość dla próbki

​$\hat{y}_i$ – wartość przewidywana przez model dla próbki


Właściwości MSE:

* MSE to średnia z kwadratów błędów - każda różnica jest podnoszona do kwadratu, co sprawia, że większe odchylenia są karane mocniej niż małe.
* Wynik MSE jest zawsze nieujemny (≥ 0). Im mniejszy, tym lepsze dopasowanie modelu, a wartość 0 oznacza, że model przechodzi dokładnie przez wszystkie punkty. Dzięki temu można go optymalizować poprzez szukanie najmniejszej wartości.

### 🧑‍🎓 Zadanie 2a

Zaimplementuj funkcję obliczającą błąd średniokwadratowy, uzupełniając brakujące miejsca w kodzie oznaczone jako `...`.

Wskazówki:

* Na tablicach można wykonywać operacje matematyczne. Są one zazwyczaj wykonywane dla każdego elementu tablicy osobno, np.:

  * dla tablicy $a$ operacja $a - 1$ oznacza odjęcie 1 od każdego elementu, tworząc tablicę $[a_0 - 1, ..., aₙ - 1]$,
  * operacja $a + b$ oznacza dodanie do każdego elementu tablicy $a$ odpowiadającego elementu tablicy $b$, tworząc $[a_0 + b_0, ..., a_n + b_n]$.
* Reszta wskazówek znajduje się w komentarzach w komórce z kodem.


```python
# W tym zadaniu korzystamy z biblioteki NumPy
import numpy as np

# Funkcja do obliczenia MSE
def mean_squared_error(y_true, y_pred):
    """
    Oblicza błąd średniokwadratowy (MSE) między wartościami rzeczywistymi a przewidywanymi.

    Parametry:
    y_true -- tablica z wartościami rzeczywistymi (np. pomiar)
    y_pred -- tablica z wartościami przewidywanymi (np. z modelu)

    Zwraca:
    mse -- wartość średniego błędu kwadratowego
    """

    # Oblicz różnicę pomiędzy wartościami rzeczywistymi i przewidywanymi
    differences = y_true - y_pred

    # Podnieś każdą różnicę do kwadratu
    squared_differences = differences ** 2

    # Oblicz średnią wartość z kwadratów różnic
    # W bibliotece NumPy (wczytanej wyżej jako np) istnieje funkcja obliczająca średnią
    mse = np.mean(squared_differences)

    # Zwróć wynik
    return mse


# Test funkcji
# Przykładowe dane (rzeczywiste i przewidywane)
y_true = np.array([1.0, 2.0, 3.0])
y_pred = np.array([1.1, 2.2, 3.5])

# Oblicz MSE
mse_result = mean_squared_error(y_true, y_pred)
print(f"Wartość MSE: {mse_result:.10f}")
```

### 🧑‍🎓 Zadanie 2b

Z tego powodu przeszukamy przestrzeń parametrów tylko w wąskim zakresie wokół wcześniej dobranych „na oko” wartości.

Przygotuj przeszukiwanie:

* Wyświetl poniżej wartości ze słownika `manual_search_points` - parametry dobrane przez Ciebie w zadaniu 1. Posłużą one jako punkt odniesienia dla metody przeglądu zupełnego.
* Zdefiniuj zakresy przeszukiwania dla każdego parametru w słowniku `grid_ranges`. Wartości pól powinny być krotkami o strukturze `(początek zakresu, koniec zakresu)`. Dokładność przeszukiwania jest regulowana przez `grid_precision` i ustawiona na `0.01`, czyli taką samą, jak w zadaniu 1.

```python
print(manual_search_points)
```

```python
# Zakres przeszukiwania współczynników
grid_ranges = {
    "a_4": (0.9 ,1.1),  # Zakres dla a_4
    "a_3": (2.35, 2.65),  # Zakres dla a_3
    "a_2": (-1.3, -1.2),  # Zakres dla a_2
    "a_1": (-10.55, -10.45),  # Zakres dla a_1
    "a_0": (4.95, 5.05)   # Zakres dla a_0
}

# Precyzja siatki przeszukiwania - krok, co jaki będą zmieniane współczynniki
grid_precision = 0.01

# Sprawdzenie spodziewanej liczby iteracji
expected_iterations = 1
for i in range(5):
    expected_iterations *= round((grid_ranges[f"a_{i}"][1] - grid_ranges[f"a_{i}"][0]) / grid_precision) + 1

print("Spodziewana liczba iteracji:", expected_iterations)
assert expected_iterations < 1e6, "Zbyt duża liczba iteracji! Zmniejsz zakres lub precyzję przeszukiwania."
```

### 🧑‍🎓 Zadanie 2c

Zaimplementuj przegląd zupełny przez uzupełnienie miejsc oznaczone jako `...` w komórkach poniżej. Komentarze w kodzie pomogą Ci wykonać zadanie.

```python
# Generowanie zakresów dla każdego współczynnika
# Słownik `ranges` powinien zawierać klucze odpowiadające współczynnikom (a_0, ..., a_4), oraz wartości będące tablicami z zakresami
# Tzn. dla krotki (0.98, 1.01) będzie to [0.98, 0.99, 1.00, 1.01], ponieważ oba końce zakresu należą do siatki.
step = int(1 / grid_precision)
ranges = {
    key: [(x/step) for x in range(int(start*step),int(end*step)+1)]
    for key, (start, end) in grid_ranges.items()
}

# Generowanie wszystkich kombinacji współczynników
# Przygotuj kombinacje tak, aby w każdej z nich współczynniki były ustawione w kolejności [a_0, a_1, ..., a_4]
# Uważaj na kolejność: podaj zakresy w kolejności a_0, a_1, …, a_4

import itertools

combinations = list(itertools.product(ranges['a_0'], ranges['a_1'],ranges['a_2'],ranges['a_3'],ranges['a_4']))
print(combinations)
```

```python
# Inicjalizacja parametrów do przechowywania najlepszego rozwiązania
best_mse = float('inf')
best_coeffs = None

# Przegląd wszystkich możliwych kombinacji a0–a4
for coeffs in tqdm(combinations, desc="Przeszukiwanie przestrzeni parametrów"):
    # Obliczanie przewidywanych wartości i MSE
    y_pred = sum(c * x**i for i, c in enumerate(coeffs))
    # Wykorzystaj wcześniej stworzoną uzupełnioną funkcję liczącą błąd średniokwadratowy
    mse = mean_squared_error(y,y_pred)
    # Sprawdź, czy obecne rozwiązanie jest lepsze od najlepszego
    # znalezionego do tej pory (czyli ma mniejszy MSE)
    if mse < best_mse:
        # Jeśli tak, zaktualizuj najlepszy MSE i współczynniki
        best_mse = mse
        best_coeffs = coeffs

# Wyniki
print("Porównanie ręcznie uzyskanych współczynników z najlepszymi znalezionymi:")
for i, a in enumerate(best_coeffs):
    print(f"Współczynnik a_{i}:", f"ręcznie wybrano {manual_search_points[f'a_{i}']:.2f} -> znaleziono {a:.2f}")
print("\n")

y_manual = sum(manual_search_points[f'a_{i}'] * x**i for i in range(5))
print("Porównanie MSE:")
print(f"Ręcznie dobrane MSE: {mean_squared_error(y, y_manual):.6f}")
print(f"MSE najlepszego znalezionego dopasowania: {best_mse:.6f}")
```

```python
from IPython.display import display, HTML

# Wizualizacja najlepszego dopasowania
y_best = sum(c * x**i for i, c in enumerate(best_coeffs))

fig = go.Figure()
fig.add_trace(go.Scatter(x=x, y=y, mode='markers', name='Dane', marker=dict(size=4, color='gray')))
fig.add_trace(go.Scatter(x=x, y=y_best, mode='lines', name='Najlepszy model', line=dict(color='blue', dash='dash')))
fig.add_trace(go.Scatter(x=x, y=y_manual, mode='lines', name='Ręcznie dopasowany model', line=dict(color='orange')))

fig.update_layout(
    title="Porównanie modeli",
    height=600,
    width=1000,
    margin=dict(l=40, r=40, t=40, b=40),
    legend=dict(x=0.01, y=0.99),
)
fig.show()

sample_indices = [0, len(x)//4, len(x)//2, 3*len(x)//4, len(x)-1]

html_content = "<div aria-live='polite' style='margin-top: 20px; font-size: 16px;'>"
html_content += "<b>Porównanie predykcji na przykładowych punktach:</b><ul>"

for idx in sample_indices:
    xi = x[idx]
    yi = y[idx]
    yp_man = y_manual[idx]
    yp_best = y_best[idx]
    
    html_content += (
        f"<li>Dla x={xi:.2f}: rzeczywista wartość y={yi:.2f}. "
        f"Twój model: {yp_man:.2f}. Najlepszy model: {yp_best:.2f}.</li>"
    )

html_content += "</ul></div>"

display(HTML(html_content))
```

Takiego modelu możemy użyć do oszacowania wartości w punkcie, którego nie zmierzono podczas eksperymentu. Przewidywania modelu zawsze będą obarczone pewnym błędem, jednak nie należy się go obawiać. Zazwyczaj znajomość wartości przybliżonej do prawdziwej jest wystarczająca z punktu widzenia rozwiązywanego problemu, a model o wysokiej (choć nie idealnej) dokładności pozostaje w pełni użyteczny.

Wykonajmy najpierw **interpolację**, czyli przewidywanie nowej wartości w obrębie zakresu, dla którego znamy wartości funkcji (posiadamy dane pomiarowe). (Uwaga: w matematyce interpolacją nazywa się też szukanie funkcji przechodzącej dokładnie przez wszystkie punkty; tutaj model jest tylko w przybliżeniu dopasowany do punktów).


```python
x_inter = 0.125

# Obliczanie predykcji dla nowego punktu
new_y = sum(c * x_inter**i for i, c in enumerate(best_coeffs))

# Rysowanie wykresu
fig = go.Figure()
fig.add_trace(go.Scatter(x=x, y=y, mode='markers', name='Dane', marker=dict(size=4, color='gray')))
fig.add_trace(go.Scatter(x=x, y=y_best, mode='lines', name='Najlepszy model', line=dict(color='blue', dash='dash')))
fig.add_trace(go.Scatter(x=[x_inter], y=[new_y], mode='markers', name='Przewidywana wartość', marker=dict(size=10, color='red')))

fig.update_layout(
    title="Interpolacja - przewidywanie wewnątrz zakresu danych uczących",
    height=600,
    width=1000,
    margin=dict(l=40, r=40, t=40, b=40),
    legend=dict(x=0.7, y=0.99),
)
fig.show()

x_arr = np.array(x)
y_arr = np.array(y)

distances = np.abs(x_arr - x_inter)
closest_indices = np.argsort(distances)[:2]

x_closest_1, y_closest_1 = x_arr[closest_indices[0]], y_arr[closest_indices[0]]
x_closest_2, y_closest_2 = x_arr[closest_indices[1]], y_arr[closest_indices[1]]

html_content = "<div aria-live='polite' style='margin-top: 20px; font-size: 16px;'>"
html_content += f"<b>Szczegóły interpolacji dla nowego punktu x = {x_inter:.3f}:</b><ul>"
html_content += f"<li><b>Przewidywana wartość przez model:</b> y = {new_y:.3f}</li>"
html_content += "<li><b>Wartości dwóch najbliższych punktów w danych uczących:</b><ul>"
html_content += f"<li>Pierwszy najbliższy punkt: x = {x_closest_1:.3f}, rzeczywiste y = {y_closest_1:.3f}</li>"
html_content += f"<li>Drugi najbliższy punkt: x = {x_closest_2:.3f}, rzeczywiste y = {y_closest_2:.3f}</li>"
html_content += "</ul></li></ul></div>"

display(HTML(html_content))
```

Co jednak się stanie, gdy spróbujemy przewidywać wartości spoza znanego zakresu, czyli wykonać **ekstrapolację**?

```python
x_extr = 1.75

# Predykcje dla nowego punktu
new_y = sum(c * x_extr**i for i, c in enumerate(best_coeffs))

# Rysowanie wykresu
fig = go.Figure()
fig.add_trace(go.Scatter(x=x, y=y, mode='markers', name='Dane', marker=dict(size=4, color='gray')))
fig.add_trace(go.Scatter(x=x, y=y_best, mode='lines', name='Najlepszy model', line=dict(color='blue', dash='dash')))
fig.add_trace(go.Scatter(x=[x_extr], y=[new_y], mode='markers', name='Przewidywana wartość', marker=dict(size=10, color='red')))

fig.update_layout(
    title="Ekstrapolacja - przewidywanie wartości poza zakresem danych uczących",
    height=600,
    width=1000,
    margin=dict(l=40, r=40, t=40, b=40),
    legend=dict(x=0.7, y=0.99, xanchor='right', yanchor='top'),
)
fig.show()

# Granice zbioru uczącego dla kontekstu
min_x = np.min(x)
max_x = np.max(x)

html_content = "<div aria-live='polite' style='margin-top: 20px; font-size: 16px;'>"
html_content += f"<b>Ekstrapolacja - dla nowego punktu x = {x_extr:.3f}:</b><ul>"
html_content += f"<li><b>Przewidywana wartość przez model:</b> y = {new_y:.3f}</li>"
html_content += "<li><b>Kontekst danych uczących:</b><ul>"
html_content += f"<li>Zakres osi x w danych uczących wynosi od {min_x:.3f} do {max_x:.3f}.</li>"
html_content += f"<li>Badany punkt x = {x_extr:.3f} znajduje się poza obszarem, na którym model był trenowany.</li>"
html_content += "</ul></li>"
html_content += "</ul></div>"

display(HTML(html_content))
```
