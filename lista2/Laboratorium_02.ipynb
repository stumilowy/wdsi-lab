### Przeszukiwanie

Wiele problemów SI polega na szukaniu: najkrótszej trasy w nawigacji, rozwiązania łamigłówki, ułożenia planu działania robota czy wyboru kolejnego ruchu w grze. Algorytmy przeszukiwania nie uczą się z danych, a sprawdzają możliwe sytuacje i wybierają najlepszą opcję. Na takich metodach opierał się m.in. komputer szachowy Deep Blue, który w 1997 r. wygrał mecz z Kasparowem (ówczesnym mistrzem świata). 

Z prostym przeszukiwaniem zetknęliśmy się już w Laboratorium 1.2, gdzie metodą przeglądu zupełnego sprawdzaliśmy po kolei wszystkie kombinacje współczynników spośród zadanych wartości.

### Przestrzeń stanów
Przeszukiwanie nie odbywa się w próżni. Aby zdefiniować problem, czyli przeszukiwaną **przestrzeń stanów**, potrzebujemy czterech kluczowych elementów:

* **Zbioru stanów** – wszystkich możliwych konfiguracji problemu (np. pozycje na mapie, układ pionków).
* **Zbioru operacji** – akcji, które powodują zmianę stanu na inny.
* **Stanu początkowego** – punktu startowego poszukiwań.
* **Funkcji końca** – testu sprawdzającego, czy dany stan jest naszym celem.

Jednym z najczęstszych sposobów reprezentacji przestrzeni stanów są **grafy**. Formalnie graf to zbiór węzłów (ang. *nodes*) oraz krawędzi (ang. *edges*) łączących je ze sobą. Podstawowy podział grafów to:

1.  **Nieskierowane** – połączenie działa w obie strony (np. droga dwukierunkowa).
2.  **Skierowane** – krawędź ma określony kierunek (np. ulica jednokierunkowa).

W grafie przestrzeni stanów każdy stan to węzeł, a każda operacja to krawędź. Problem rozwiązujemy szukając ścieżki, czyli ciągu krawędzi od stanu początkowego do stanu końcowego. 

### Jak szukać w grafie?
Najprostszą strategią jest metoda **brute-force** (przeszukiwanie wszystkich ścieżek - przegląd zupełny), jednak jest ona niewydajna – liczba obliczeń rośnie wykładniczo wraz z długością ścieżek. Algorytmy przeszukiwania grafu, które omówimy sprawdzają każdy stan najwyżej raz, więc koszt zależy od liczby stanów, a nie ścieżek. Algorytmy te dzielimy na:

* **Nieinformowane** – szukamy "na ślepo": w każdym stanie znamy jego sąsiadów i umiemy rozpoznać cel, ale nie wiemy, który kierunek nas do niego przybliży (np. błądzenie po nieznanym sklepie).
* **Informowane** – korzystamy z heurystyki, czyli reguły, która szacuje jak daleko od celu jesteśmy (np. wiemy, że lody zawsze są w zamrażarkach, a te są zazwyczaj na końcu sklepu).

#### Algorytmy nieinformowane
W tej sekcji skupimy się na dwóch fundamentalnych metodach:

1.  **DFS (Depth-First Search) – Przeszukiwanie w głąb**
 wybiera jedną ścieżkę i podąża nią do końca. Jeśli trafi na ślepy zaułek, cofa się (backtracking) i sprawdza inną gałąź. Zużywa niedużo pamięci, ale nie gwarantuje znalezienia najkrótszej drogi.
    
2.  **BFS (Breadth-First Search) – Przeszukiwanie wszerz** 
działa "warstwami" – najpierw odwiedza wszystkich bezpośrednich sąsiadów, potem sąsiadów sąsiadów. Gwarantuje znalezienie ścieżki o najmniejszej liczbie kroków, ale zużywa znacznie więcej pamięci niż DFS.

Oba algorytmy zapamiętują odwiedzone stany i nie wracają do nich. Bez tego w grafie z cyklami mogłyby krążyć w nieskończoność.

### Przeszukiwanie stanu gry i algorytm Minimax
W grach o **stałej sumie** (gdzie zysk jednego gracza jest stratą drugiego; zazwyczaj nazywanych grami o **sumie zerowej**), stosujemy algorytm **Minimax**. Opiera się on na definiowaniu `wartości stanu gry`.

* Wartość stanu to końcowy wynik gry z punktu widzenia gracza MAX (np. +1 wygrana, 0 remis, −1 przegrana, albo różnica punktów graczy), jaki zostanie osiągnięty, jeśli od tego stanu obaj gracze będą grać optymalnie.
* Minimax wyznacza ją, budując drzewo gry. Drzewo to graf bez cykli z wyróżnionym węzłem startowym – korzeniem; węzły bez dzieci to liście. Korzeń to stan bieżący, dzieci węzła to stany po każdym możliwym ruchu, a liście to stany końcowe. Wartości liści znamy (to wynik gry).
*  Następnie przenosimy wartości w górę, poziom po poziomie. Jeden gracz (**MAX**) stara się w swoim ruchu tę wartość zmaksymalizować, drugi (**MIN**) stara się ją zminimalizować.

W grach takich jak szachy, gdzie pełne drzewo gry jest zbyt duże, budujemy drzewo możliwych ruchów do określonej głębokości, a następnie szacuje ich wartość funkcją oceny, np. przewagą wartości figur. Te wartości przenosi się w górę, zakładając, że obaj gracze grają optymalnie. W takim przypadku, wynik algorytmu jest przybliżeniem.

W Othello 1D (zadanie w dalszej części laboratorium) na końcu gry suma pionków obu graczy zawsze wynosi 8, więc pionek zdobyty przez jednego gracza to pionek stracony przez drugiego.

## Implementacja Algorytmu Minimax

Uzupełnij pięć brakujących fragmentów kodu wewnątrz funkcji `minimax`. Funkcja minimax jest rekurencyjna, czyli wywołuje samą siebie. Łańcuch wywołań kończy się, gdy gra się skończy (krok 1). Wtedy funkcja zwraca wynik zamiast wywoływać się dalej. Taki warunek jest konieczny, aby funkcja nie wywoływała się bez końca.

1. Sprawdź, czy gra się skończyła i zwróć jej wynik.
2. W gałęzi gracza maksymalizującego wywołaj rekurencyjnie funkcję dla nowej planszy i tury przeciwnika; `maximizing_color` się nie zmienia.
3. Zachowaj największy z dotychczasowych wyników. 
4. W gałęzi gracza minimalizującego wywołaj rekurencyjnie funkcję dla nowej planszy i tury przeciwnika. 
5. Zachowaj najmniejszy z dotychczasowych wyników.

**Wskazówki:**
- `is_game_over(board)` zwraca `True`, gdy gra dobiegła końca.
- `calculate_score(board, maximizing_color)` zwraca różnicę między wynikiem gracza maksymalizującego i minimalizującego. Liczba jest dodatnia, gdy gracz maximizing_color ma przewagę.
- `make_move(board, move, color)` zwraca krotkę `(new_board, flipped_indices)`.
  Wywołanie rekurencyjne powinno odbyć się na nowej planszy (`new_board`).
- Kiedy wywołujesz funkcję rekurencyjnie, następuje tura **przeciwnika**.
- Graczowi maksymalizującemu zależy na jak najwyższej, czy jak najniższej wartości best_score?

```python
def minimax(board, current_turn_color, maximizing_color):
    valid_moves = get_valid_moves(board)
    opponent_color = 'O' if current_turn_color == 'X' else 'X'
    
    # Sprawdź czy gra została zakończona wykorzystując funkcję is_game_over(). 
    # Jeżeli tak, zwróć wynik dla gracza maksymalizującego (maximizing_color) 
    # Wynik obliczysz korzystając z funkcji calculate_score().
    if is_game_over(board):
        return calculate_score(board, maximizing_color)

    if current_turn_color == maximizing_color: # Gracz maksymalizujący
        # Najmniejsza mozliwa wartość best_score
        best_score = -math.inf

        # Obliczamy wynik dla każdego dostępnego ruchu
        for move in valid_moves:
            new_board, _ = make_move(board, move, current_turn_color)
            
            # Wywołaj funkcję minmax() dla tury przeciwnika.
            score = minimax(new_board, opponent_color, maximizing_color)
            
            # Zaktualizuj wartość best_score.
            best_score = max(best_score,score)
            
        return best_score
        
    else: # Gracz minimalizujący
        # Największa mozliwa wartość best_score
        best_score = math.inf

        # Obliczamy wynik dla każdego dostępnego ruchu
        for move in valid_moves:
            new_board, _ = make_move(board, move, current_turn_color)
            
            # Wywołaj funkcję minmax() dla tury przeciwnika.
            score = minimax(new_board,opponent_color,maximizing_color)
            
            # Zaktualizuj wartość best_score.
            best_score = min(best_score, score)
            
        return best_score
```

## Agent Minimax

Uzupełnij funkcję implementującą gracza minimax - `minimax_agent`. Wywołaj `minimax`, aby ocenić stan gry, a następnie zaktualizuj zmienne `best_score` i `best_move`.

Agent jest graczem maksymalizującym. Dla każdego ruchu wykonaj go na kopii planszy i oceń powstały stan wywołaniem `minimax`. Po ruchu agenta następuje kolej przeciwnika.

```python
def minimax_agent(board, player_color, valid_moves):
    opponent_color = 'O' if player_color == 'X' else 'X'
    best_score = -math.inf
    best_move = valid_moves[0]
    
    # Dla każdego dostępnego ruchu
    for move in valid_moves:
        new_board, _ = make_move(board, move, player_color)
        # Wywołaj funkcję minimax() i oceń stan gry
        score = minimax(new_board, opponent_color, player_color)
        # Jeżeli wynik jest najlepszy z dotychczas widzianych
        # zaktualizuj najwyższy wynik i najlepszy ruch
        if score > best_score:
            best_score = score
            best_move = move
        
    return best_move
```
