# Laboratorium kursu Wstęp do Sztucznej Inteligencji
## Systemy logiczne i wnioskowanie

Systemy logiczne (regułowe) to jeden z najstarszych nurtów SI (tzw. **SI symboliczna**). Wiedzę zapisujemy w nich jawnie, w postaci czytelnej dla człowieka, a rozumowania można prześledzić krok po kroku. Składają się one z:
* **bazy wiedzy** – faktów i reguł JEŻELI… TO… 
* **mechanizmu wnioskowania**, który z tej wiedzy wyprowadza nowe fakty.  

W tym laboratorium poznasz podstawowe metody wnioskowania. **Dedukcja** prowadzi od reguł i faktów do wniosków; jeśli reguły i fakty są prawdziwe, wniosek też jest prawdziwy. **Indukcja** prowadzi od wielu przykładów do reguły ogólnej, która może mieć wyjątki. **Abdukcja** szuka założenia, które wyjaśnia obserwację, a jej wynik to hipoteza. Zobaczysz, jak komputer wnioskuje z reguł zapisanych przez człowieka, dowiesz się skąd brać reguły, gdy nikt ich nie zna, i nauczysz się odkrywać je w danych.

*Choć systemy regułowe nadal są używane, np. w bankowości i ubezpieczeniach, popularne dziś duże modele językowe działają inaczej: ich wiedza jest zapisana niejawnie w ogromnej liczbie parametrów, a odpowiedź tworzą przez przewidywanie kolejnych słów.*

## Reguły logiczne

*Uwaga - przykłady są w języku, który moglibyśmy określić mianem 'pseudo-prologu' - składni wzorowanej na Prologu. Program w Prologu nie jest listą poleceń, tylko zbiorem faktów i reguł (wspólnie nazywanych **klauzulami**), czyli bazą wiedzy. Obliczenie polega na zadaniu pytania, na które system odpowiada, wnioskując z bazy.*

Bazę wiedzy (Knowledge Base, KB) można traktować jako zbiór dwóch typów informacji:

1.  **Fakty (Facts):** To proste, bezwarunkowe stwierdzenia o świecie, które uznajemy za prawdziwe.
      * `Pesa produkuje pociągi`
      * `We Wrocławiu pada deszcz`
      * `Jan jest tatą Ani`
    
    Powyższe fakty możemy zapisać przy pomocy predykatów. Predykat to nazwa własności lub relacji, np. `jest_tata`. Po podaniu argumentów (obiektów) otrzymujemy stwierdzenie, które jest prawdziwe albo fałszywe, np. `jest_tata(jan, ania)`, czyli „Jan jest tatą Ani”.

2.  **Reguły (Rules):** To stwierdzenia warunkowe (implikacje), które opisują zależności w świecie. Mówią: "JEŚLI coś jest prawdą, TO coś innego również jest prawdą".

    W Prologu reguły zapisujemy "od tyłu". Zamiast pisać `JEŚLI P TO Q`, piszemy `Q :- P` (głowa reguły znajduje się z lewej strony, ciało z prawej).  Zapis `Głowa :- Ciało` odpowiada implikacji logicznej $Ciało \Rightarrow Głowa$. Ciało to **poprzednik** (przesłanka), a głowa to **następnik** (wniosek).

    `jest_mokro(wroclaw) :- pada_deszcz(wroclaw).`

    `zmokne(wroclaw) :- jest_mokro(wroclaw), wyjde_z_domu(wroclaw).`

    * **jest_mokro(wroclaw)** jest wnioskiem (głową), który jest prawdziwy gdy przesłanka **pada_deszcz(wroclaw)** jest spełniona.
    * Przecinek `,` w ciele reguły oznacza logiczne "**i**" (koniunkcję), średnik `;` zaś logiczne **lub** (alternatywę). Alternatywę możemy też zapisać przez kilka reguł o tej samej głowie. Wystarczy wtedy, że zadziała jedna z nich.

    Powyższy zapis jednak nie pozwala tworzyć uogólnionych, aplikowalnych reguł. 
    Do tego służą **zmienne**. W Prologu zmienne zaczynają się **wielką literą** (`X`, `Y`, `Osoba`), a obiekty **małą literą** (`wroclaw`, `jan`). Specjalna zmienna `_` oznacza cokolwiek, wartość nas nie interesuje.

    Załóżmy, że we Wrocławiu pada deszcz `pada_deszcz(wroclaw).`.
    Chcemy wyrazić regułę: "Gdy gdzieś pada, to jest tam mokro", czyli w X jest mokro, JEŚLI w X pada deszcz: 
    `jest_mokro(X) :- pada_deszcz(X).`

    Czasem jednak musimy dookreślić wykluczenia. Załóżmy, że Asia i Janek studiują Sztuczną Inteligencję. Możemy stworzyć fakty `studiuje(asia, si).` i `studiuje(janek, si).`. 
    Naturalna też wydaje się reguła mówiąca, że są kolegami z kierunku: `koledzy_z_kierunku(X, Y) :- studiuje(X, K), studiuje(Y, K), X \= Y.`
    Zmienna `K` występuje dwa razy, więc obie osoby muszą studiować **ten sam** kierunek. Poprzez `X \= Y` zapisujemy, że X nie może być równe Y, czyli osoba nie może być sama sobie kolegą. Bez niego, możliwe byłoby wywołanie `studiuje(asia, K), studiuje(asia, K)` i w efekcie dojście do wniosku, że wyrażenie `koledzy_z_kierunku(asia, asia)` jest prawdą.


### Wnioskowanie w przód (Forward Chaining)

Wnioskowanie to proces, w którym system SI (nazywany *silnikiem wnioskującym*) używa faktów i reguł, aby **wyprowadzić nowe fakty**. Nowe czyli takie, których nie wpisaliśmy do bazy wprost, ale które z niej wynikają. Taki rodzaj wnioskowania, od reguł i faktów do pewnych wniosków, to **dedukcja**. 

Główny mechanizm to **Modus ponens**:

1.  Mamy regułę `głowa :- ciało` (*Jeśli ciało, to głowa*)
2.  Wiemy, że `ciało` jest prawdziwe - jest to wpisany fakt, lub system wcześniej to udowodnił.
3.  Z naszej reguły (*Jeśli ciało, to głowa*), system może wywnioskować, że `głowa` również jest prawdziwa.

Tak jak w poprzednim przykładzie - jeśli wiemy, że `jest_mokro(X) :- pada_deszcz(X).` i `pada_deszcz(wroclaw).` jest prawdziwe, możemy wywnioskować, że `jest_mokro(wroclaw)` jest prawdziwe. Jeżeli dodamy regułę `jest_ślisko(X) :- jest_mokro(X).`, to system, łącząc reguły w łańcuch może wywnioskować, że `jest_ślisko(wroclaw)` jest prawdziwe. Takie wnioskowanie (od znanych faktów, przez reguły, do nowych faktów), nazywamy **wnioskowaniem w przód**.

### Wnioskowanie wstecz (Backward Chaining)

Możemy też wnioskować zaczynając od pytania (celu) i szukać reguł, które pozwolą go udowodnić. To **wnioskowanie wstecz** (*backward chaining*). W tym rodzaju wnioskowania sprawdzamy tylko to, co jest potrzebne do odpowiedzi na zadane pytanie. Kiedy zadajemy systemowi pytanie, np. `?- jest_mokro(wroclaw).` (Czy we Wrocławiu jest mokro?), system regułowy zadziała "wstecz":

**Cel:** Udowodnić `jest_mokro(wroclaw)`:

* System znajduje pasującą regułę: `jest_mokro(X) :- pada_deszcz(X).`.
* System dopasowuje zmienne. Aby głowa (lewa strona reguły) pasowała do celu, `X` musi być równe `wroclaw`.
* System musi teraz udowodnić *ciało* tej reguły, czyli `pada_deszcz(wroclaw)`
* System przeszukuje bazę wiedzy i znajduje fakt: `pada_deszcz(wroclaw).`
* Ciało jest prawdziwe, zatem głowa też jest prawdziwa. Cel został udowodniony.

**Odpowiedź:** `true` (Tak), czyli "We Wrocławiu jest mokro".


Wnioskowanie w przód i wstecz to dwa sposoby prowadzenia **dedukcji**, ponieważ oba dają pewne wnioski z faktów i reguł (różnią się jednak punktem startowym - fakty lub pytanie). Abdukcja z kolei pozwala *założyć* coś, czego w bazie nie ma, dlatego daje tylko hipotezę.

Przykład 1:
```
near(cukiernia, c16).
near(cukiernia, a1).
near(student, c16).  

near_sym(X, Y) :- near(X, Y).
near_sym(X, Y) :- near(Y, X).

reachable(X, Y) :- near_sym(X, Y).
reachable(X, Y) :- near_sym(X, Z), reachable(Z, Y).

can_visit_between(From, To, Place) :-
    reachable(From, Place),
    reachable(Place, To).

?- can_visit_between(c16, a1, cukiernia).
```



Przykład 2:
```
on_sky(wroclaw, moon).
on_sky(wroclaw, saturn).
on_sky(wroclaw, altair).

type(moon, satellite).
type(saturn, planet).
type(altair, star).

~on_sky(wroclaw, sun). 
~explode(altair).

abducible(phase(_, _)).

bright_sky(X) :- on_sky(X, sun).
bright_sky(X) :- on_sky(X, moon), phase(moon, full).
bright_sky(X) :- on_sky(X, Y), supernova(Y).

supernova(X) :- type(X, star), explode(X).

?- bright_sky(wroclaw). 
```

W przykładzie 2 pojawiają się nowe elementy zapisu: 
* `~fakt.` to **jawne zaprzeczenie**, czyli to *nie* jest prawda 
* `_` to zmienna oznaczająca „cokolwiek” 

`abducible(phase(_, _)).` oznacza, że faktów o fazie (`phase`) nie znamy, ale **wolno je założyć**, jeśli pozwoli to udowodnić cel. Szukanie założenia wyjaśniającego obserwację nazywamy **abdukcją**. Wniosek z abdukcji jest tylko hipotezą i jest prawdziwy pod warunkiem, że prawdziwe jest założenie. Odpowiedź może więc brzmieć "celu nie da się udowodnić" albo "cel jest prawdziwy, jeśli założymy ...".

## Gemini Argon
### Przykład 1: Wnioskowanie dedukcyjne (SLD-rezolucja)

**Wynik:** `true` (Prawda)

**Tok rozumowania:**
Celem głównym jest udowodnienie zapytania `can_visit_between(c16, a1, cukiernia)`.

1. **Rozwinięcie celu głównego:**
Zgodnie z regułą `can_visit_between(From, To, Place)`, podstawiamy `From = c16`, `To = a1`, `Place = cukiernia`. Otrzymujemy dwa podcele, które muszą zostać spełnione jednocześnie:
* **Podcel 1:** `reachable(c16, cukiernia)`
* **Podcel 2:** `reachable(cukiernia, a1)`


2. **Dowodzenie Podcelu 1 (`reachable(c16, cukiernia)`):**
* Stosujemy pierwszą regułę dla relacji osiągalności: `reachable(X, Y) :- near_sym(X, Y)`. Podstawiamy `X = c16`, `Y = cukiernia`, co daje nowy cel: `near_sym(c16, cukiernia)`.
* Próbujemy pierwszej reguły symetrii: `near_sym(X, Y) :- near(X, Y)`. Szukamy faktu `near(c16, cukiernia)` w bazie wiedzy. Brak takiego faktu (porażka i nawrót / *backtracking*).
* Próbujemy drugiej reguły symetrii: `near_sym(X, Y) :- near(Y, X)`. Dla `X = c16`, `Y = cukiernia` szukamy faktu `near(cukiernia, c16)`.
* Fakt `near(cukiernia, c16)` istnieje w bazie (linia 1). **Podcel 1 zostaje udowodniony.**


3. **Dowodzenie Podcelu 2 (`reachable(cukiernia, a1)`):**
* Stosujemy pierwszą regułę: `reachable(X, Y) :- near_sym(X, Y)` dla `X = cukiernia`, `Y = a1`. Cel zmienia się na `near_sym(cukiernia, a1)`.
* Stosujemy pierwszą regułę symetrii: `near_sym(X, Y) :- near(X, Y)`, co sprowadza się do sprawdzenia faktu `near(cukiernia, a1)`.
* Fakt `near(cukiernia, a1)` istnieje w bazie (linia 2). **Podcel 2 zostaje udowodniony.**


4. **Konkluzja:** Oba podcele są prawdziwe, zatem zapytanie `can_visit_between(c16, a1, cukiernia)` zwraca wartość **`true`**.

### Przykład 2: Wnioskowanie abdukcyjne (ALP – Abductive Logic Programming)

**Wynik:**

* **Klasyczny Prolog (dedukcja):** `false` (brak faktu `phase(moon, full)` w bazie).
* **Z abdukcją:** `true` pod warunkiem przyjęcia hipotezy abdukcyjnej $\Delta = \{\text{phase(moon, full)}\}$.

**Tok rozumowania:**
Celem jest wyjaśnienie obserwacji `bright_sky(wroclaw)`. W programie zadeklarowano `abducible(phase(_, _))`, co oznacza, że predykat `phase` może być traktowany jako założenie (hipoteza), jeśli nie ma go wprost w bazie faktów i nie jest sprzeczny z więzami integralności.

1. **Ścieżka 1 (Słońce):**
* Reguła 1: `bright_sky(X) :- on_sky(X, sun).` (podstawiamy `X = wroclaw`).
* Podcel: `on_sky(wroclaw, sun)`.
* W bazie wiedzy znajduje się jawna negacja `~on_sky(wroclaw, sun)` (oraz brak faktu pozytywnego). Ścieżka kończy się **porażką**.


2. **Ścieżka 2 (Księżyc i abdukcja):**
* Reguła 2: `bright_sky(X) :- on_sky(X, moon), phase(moon, full).` (dla `X = wroclaw`).
* Podcel 2a: `on_sky(wroclaw, moon)`. Fakt znajduje się w bazie (linia 1) – **sukces**.
* Podcel 2b: `phase(moon, full)`. Tego faktu nie ma w bazie wiedzy, nie ma też reguł pozwalających go wyprowadzić. Ponieważ jednak zadeklarowano `abducible(phase(_, _))` i w bazie nie ma jawnej negacji `~phase(moon, full)`, dodajemy `phase(moon, full)` do zbioru hipotez wyjaśniających $\Delta$. Ścieżka kończy się **sukcesem warunkowym**.


3. **Ścieżka 3 (Supernowa – weryfikacja alternatyw):**
* Reguła 3: `bright_sky(X) :- on_sky(X, Y), supernova(Y).` (dla `X = wroclaw`).
* Szukamy obiektów `Y` spełniających `on_sky(wroclaw, Y)`. Baza zwraca kolejno: `moon`, `saturn`, `altair`.
* Rozwijamy `supernova(Y)` zgodnie z regułą: `supernova(Y) :- type(Y, star), explode(Y).`
* Dla `Y = moon` oraz `Y = saturn`: warunek `type(Y, star)` zawodzi (w bazie mają typy `satellite` oraz `planet`).
* Dla `Y = altair`: warunek `type(altair, star)` jest spełniony (linia 6). Przechodzimy do podcelu `explode(altair)`. W bazie widnieje jawna negacja `~explode(altair)` (oraz `explode` nie jest predykatem abdukcyjnym). Ścieżka kończy się **porażką**.


4. **Konkluzja:** Jedynym sposobem na udowodnienie `bright_sky(wroclaw)` jest wykorzystanie Ścieżki 2 poprzez abdukcję. Zapytanie jest spełnione przy założeniu hipotezy **`phase(moon, full)`** (na niebie we Wrocławiu jest jasno, ponieważ świeci księżyc i zakładamy, że jest w pełni).
